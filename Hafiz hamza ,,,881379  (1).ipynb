## Create a class-based bank account system. Handle deposits and withdrawals using custom exceptions for invalid transaction amounts and insufficient balance.

```python
class InvalidAmountError(Exception):
    pass


class InsufficientBalanceError(Exception):
    pass


class BankAccount:
    def __init__(self, balance=0):
        self.balance = balance

    def deposit(self, amount):
        if amount <= 0:
            raise InvalidAmountError("Deposit amount must be greater than 0.")
        self.balance = round(self.balance + amount, 2)
        print("Deposited:", round(amount, 2), "Balance:", self.balance)

    def withdraw(self, amount):
        if amount <= 0:
            raise InvalidAmountError("Withdrawal amount must be greater than 0.")
        if amount > self.balance:
            raise InsufficientBalanceError("Insufficient balance.")
        self.balance = round(self.balance - amount, 2)
        print("Withdrawn:", round(amount, 2), "Balance:", self.balance)


account = BankAccount(1000)

try:
    dep = float(input("Enter deposit amount: "))
    account.deposit(dep)

    wd = float(input("Enter withdrawal amount: "))
    account.withdraw(wd)
except InvalidAmountError as e:
    print("Transaction error:", e)
except InsufficientBalanceError as e:
    print("Transaction error:", e)
except ValueError:
    print("Please enter a valid number.")
```

## Create an e-commerce order system that handles invalid products, invalid quantities, insufficient stock, and invalid payments. Use custom exceptions for business-rule failures.

```python
class InvalidProductError(Exception):
    pass


class InvalidQuantityError(Exception):
    pass


class InsufficientStockError(Exception):
    pass


class InvalidPaymentError(Exception):
    pass


products = {
    "laptop": {"price": 80000, "stock": 5},
    "mouse": {"price": 1500, "stock": 20},
    "keyboard": {"price": 3000, "stock": 10}
}

valid_payments = ["cash", "card", "bank"]

try:
    item = input("Enter product name (laptop/mouse/keyboard): ").lower()
    if item not in products:
        raise InvalidProductError("Product does not exist.")

    qty = int(input("Enter quantity: "))
    if qty <= 0:
        raise InvalidQuantityError("Quantity must be greater than 0.")
    if qty > products[item]["stock"]:
        raise InsufficientStockError("Not enough stock available.")

    payment = input("Enter payment method (cash/card/bank): ").lower()
    if payment not in valid_payments:
        raise InvalidPaymentError("Invalid payment method.")

    total = round(products[item]["price"] * qty, 2)
    products[item]["stock"] = products[item]["stock"] - qty
    print("Order placed successfully. Total amount:", total)
    print("Remaining stock:", products[item]["stock"])
except InvalidProductError as e:
    print("Order failed:", e)
except InvalidQuantityError as e:
    print("Order failed:", e)
except InsufficientStockError as e:
    print("Order failed:", e)
except InvalidPaymentError as e:
    print("Order failed:", e)
except ValueError:
    print("Please enter a valid number for quantity.")
```

## ### University Course Registration System

### Goal:
Create a program that allows students to register for university courses.

### Steps:
- Store students and courses.
- Check whether the student exists.
- Check whether the course exists.
- Prevent duplicate registration.
- Check course capacity.
- Check prerequisites.
- Create and handle suitable custom exceptions.

### Example:
Student ID: 102
Course: Python
→ Registration successful

```python
class StudentNotFoundError(Exception):
    pass


class CourseNotFoundError(Exception):
    pass


class DuplicateRegistrationError(Exception):
    pass


class CourseFullError(Exception):
    pass


class PrerequisiteError(Exception):
    pass


students = {
    101: {"name": "Ali", "completed": [], "enrolled": []},
    102: {"name": "Sara", "completed": ["Programming Basics"], "enrolled": []},
    103: {"name": "Bilal", "completed": ["Programming Basics", "Python"], "enrolled": ["Python"]}
}

courses = {
    "Programming Basics": {"capacity": 2, "enrolled": 2, "prereq": None},
    "Python": {"capacity": 2, "enrolled": 1, "prereq": "Programming Basics"},
    "Databases": {"capacity": 1, "enrolled": 0, "prereq": "Python"}
}

try:
    sid = int(input("Enter Student ID: "))
    if sid not in students:
        raise StudentNotFoundError("Student not found.")

    cname = input("Enter Course name: ")
    if cname not in courses:
        raise CourseNotFoundError("Course not found.")

    student = students[sid]
    course = courses[cname]

    if cname in student["enrolled"]:
        raise DuplicateRegistrationError("Student is already registered for this course.")

    if course["enrolled"] >= course["capacity"]:
        raise CourseFullError("Course capacity is full.")

    prereq = course["prereq"]
    if prereq is not None and prereq not in student["completed"]:
        raise PrerequisiteError("Missing prerequisite: " + prereq)

    student["enrolled"].append(cname)
    course["enrolled"] = course["enrolled"] + 1
    print("Student ID:", sid, "Course:", cname, "-> Registration successful")
except StudentNotFoundError as e:
    print("Registration failed:", e)
except CourseNotFoundError as e:
    print("Registration failed:", e)
except DuplicateRegistrationError as e:
    print("Registration failed:", e)
except CourseFullError as e:
    print("Registration failed:", e)
except PrerequisiteError as e:
    print("Registration failed:", e)
except ValueError:
    print("Please enter a valid number for Student ID.")
```

## ### Hospital Appointment System

### Goal:
Create a program that manages hospital appointments.

### Steps:
- Store patients and doctors.
- Validate patient and doctor information.
- Check doctor availability.
- Prevent duplicate appointment times.
- Create custom exceptions for important business errors.
- Handle invalid input without crashing.

### Example:
Patient ID: P102
Doctor: D05
Time: 10:30 AM
→ Appointment booked successfully

```python
class PatientNotFoundError(Exception):
    pass


class DoctorNotFoundError(Exception):
    pass


class DoctorUnavailableError(Exception):
    pass


class DuplicateAppointmentError(Exception):
    pass


patients = {
    "P101": {"name": "Ali"},
    "P102": {"name": "Sara"},
    "P103": {"name": "Bilal"}
}

doctors = {
    "D01": {"name": "Dr. Ahmed", "available_times": ["09:00 AM", "10:00 AM", "11:00 AM"]},
    "D05": {"name": "Dr. Ayesha", "available_times": ["10:30 AM", "12:00 PM", "02:00 PM"]}
}

appointments = [
    {"patient_id": "P101", "doctor_id": "D05", "time": "12:00 PM"}
]

try:
    pid = input("Enter Patient ID: ").strip().upper()
    if pid not in patients:
        raise PatientNotFoundError("Patient not found.")

    did = input("Enter Doctor ID: ").strip().upper()
    if did not in doctors:
        raise DoctorNotFoundError("Doctor not found.")

    slot = input("Enter Appointment Time (e.g. 10:30 AM): ").strip().upper()
    if slot not in doctors[did]["available_times"]:
        raise DoctorUnavailableError("Doctor is not available at this time.")

    for app in appointments:
        if app["doctor_id"] == did and app["time"] == slot:
            raise DuplicateAppointmentError("This appointment slot is already booked.")
        if app["patient_id"] == pid and app["time"] == slot:
            raise DuplicateAppointmentError("Patient already has an appointment at this time.")

    appointments.append({"patient_id": pid, "doctor_id": did, "time": slot})
    print("Patient ID:", pid, "Doctor:", did, "Time:", slot, "-> Appointment booked successfully")
except PatientNotFoundError as e:
    print("Booking failed:", e)
except DoctorNotFoundError as e:
    print("Booking failed:", e)
except DoctorUnavailableError as e:
    print("Booking failed:", e)
except DuplicateAppointmentError as e:
    print("Booking failed:", e)
except Exception as e:
    print("Invalid input or error:", e)
```

## ### Digital Wallet System

### Goal:
Create a class-based digital wallet application.

### Steps:
- Validate wallet ID and PIN.
- Handle inactive accounts.
- Validate transfer amounts.
- Check balance.
- Enforce a transfer limit.
- Create multiple custom exceptions.
- Make sure failed transfers do not change wallet balances.

### Example:
Transfer: Rs. 2,000
→ Transfer successful

```python
# class WalletNotFoundError(Exception):
    pass


class InvalidPinError(Exception):
    pass


class InactiveWalletError(Exception):
    pass


class InvalidAmountError(Exception):
    pass


class InsufficientBalanceError(Exception):
    pass


class TransferLimitExceededError(Exception):
    pass


class DigitalWallet:
    def __init__(self, wallet_id, pin, balance, active=True, transfer_limit=5000):
        self.wallet_id = wallet_id
        self.pin = pin
        self.balance = balance
        self.active = active
        self.transfer_limit = transfer_limit

    def verify_pin(self, entered_pin):
        if entered_pin != self.pin:
            raise InvalidPinError("Invalid PIN.")

    def check_active(self):
        if not self.active:
            raise InactiveWalletError("Wallet is inactive.")

    def transfer(self, receiver, amount):
        self.check_active()
        receiver.check_active()

        if amount <= 0:
            raise InvalidAmountError("Transfer amount must be greater than 0.")
        if amount > self.transfer_limit:
            raise TransferLimitExceededError("Transfer amount exceeds limit.")
        if amount > self.balance:
            raise InsufficientBalanceError("Insufficient balance.")

        self.balance = round(self.balance - amount, 2)
        receiver.balance = round(receiver.balance + amount, 2)
        print("Transfer: Rs.", round(amount, 2), "-> Transfer successful")
        print("New Balance: Rs.", self.balance)


wallets = {
    "W101": DigitalWallet("W101", "1234", 10000.0, active=True, transfer_limit=5000.0),
    "W102": DigitalWallet("W102", "5678", 3000.0, active=True, transfer_limit=5000.0),
    "W103": DigitalWallet("W103", "9999", 2000.0, active=False, transfer_limit=5000.0)
}

try:
    sender_id = input("Enter your Wallet ID: ").strip().upper()
    if sender_id not in wallets:
        raise WalletNotFoundError("Sender wallet not found.")

    pin = input("Enter PIN: ").strip()
    wallets[sender_id].verify_pin(pin)

    receiver_id = input("Enter Receiver Wallet ID: ").strip().upper()
    if receiver_id not in wallets:
        raise WalletNotFoundError("Receiver wallet not found.")

    amount = float(input("Enter transfer amount: "))
    wallets[sender_id].transfer(wallets[receiver_id], amount)
except WalletNotFoundError as e:
    print("Transfer failed:", e)
except InvalidPinError as e:
    print("Transfer failed:", e)
except InactiveWalletError as e:
    print("Transfer failed:", e)
except InvalidAmountError as e:
    print("Transfer failed:", e)
except TransferLimitExceededError as e:
    print("Transfer failed:", e)
except InsufficientBalanceError as e:
    print("Transfer failed:", e)
except ValueError:
    print("Please enter a valid number for amount.")
```

## ### Retail Inventory and Sales System

### Task:
Create a menu-driven inventory and sales application.

The program should allow the user to:
- Add stock
- View inventory
- Sell products
- Search for products

Handle invalid input, missing products, invalid quantities, and insufficient stock. Use `raise`, custom exceptions, `else`, and `finally` where appropriate.

The application should continue running after recoverable errors.

```python
class ItemError(Exception):
    pass


class StockError(Exception):
    pass


stock = {"pen": [20.0, 10], "book": [150.0, 5]}

while True:
    print("\n1.Add 2.View 3.Sell 4.Search 5.Exit")
    c = input("Choice: ").strip()
    if c == "5":
        break
    try:
        if c == "1":
            n = input("Name: ").strip().lower()
            q = int(input("Qty: "))
            if q <= 0:
                raise StockError("Bad qty")
            if n in stock:
                stock[n][1] = stock[n][1] + q
            else:
                p = float(input("Price: "))
                if p <= 0:
                    raise ValueError("Bad price")
                stock[n] = [round(p, 2), q]
        elif c == "2":
            for n in stock:
                print(n, round(stock[n][0], 2), stock[n][1])
        elif c == "3":
            n = input("Name: ").strip().lower()
            if n not in stock:
                raise ItemError("Not found")
            q = int(input("Qty: "))
            if q <= 0 or q > stock[n][1]:
                raise StockError("Invalid or low stock")
            stock[n][1] = stock[n][1] - q
            print("Total:", round(stock[n][0] * q, 2))
        elif c == "4":
            n = input("Name: ").strip().lower()
            if n not in stock:
                raise ItemError("Not found")
            print(n, round(stock[n][0], 2), stock[n][1])
        else:
            print("Invalid choice")
    except (ItemError, StockError, ValueError) as e:
        print("Error:", e)
    else:
        print("Success")
    finally:
        print("Step done")
```

```python
class CustomerError(Exception):
    pass


class RestaurantError(Exception):
    pass


class MenuItemError(Exception):
    pass


class StockError(Exception):
    pass


class MinimumOrderError(Exception):
    pass


class DeliveryAreaError(Exception):
    pass


class PaymentError(Exception):
    pass


customers = {
    "Ali": {"area": "Gulshan", "balance": 2000.0}
}

restaurants = {
    "ABC Foods": {
        "areas": ["Gulshan", "DHA", "Tariq Road"],
        "min_order": 500.0,
        "menu": {
            "Burger": {"price": 450.0, "stock": 10},
            "Pizza": {"price": 1200.0, "stock": 4},
            "Fries": {"price": 250.0, "stock": 15}
        }
    }
}

orders = []

try:
    reg = input("Register new customer? (yes/no): ").strip().lower()
    if reg == "yes":
        cname = input("Enter customer name: ").strip().title()
        if cname == "":
            raise CustomerError("Customer name cannot be empty.")
        carea = input("Enter delivery area: ").strip().title()
        cbal = float(input("Enter wallet balance: "))
        if cbal < 0:
            raise PaymentError("Balance cannot be negative.")
        customers[cname] = {"area": carea, "balance": round(cbal, 2)}
    else:
        cname = input("Enter existing customer name: ").strip().title()
        if cname not in customers:
            raise CustomerError("Customer not found.")

    rname = input("Select restaurant (ABC Foods): ").strip().title()
    if rname not in restaurants:
        raise RestaurantError("Restaurant not found.")

    rest = restaurants[rname]
    cust = customers[cname]

    if cust["area"] not in rest["areas"]:
        raise DeliveryAreaError("Delivery not available in customer area.")

    print("Menu:")
    for item_name, info in rest["menu"].items():
        print("-", item_name, "| Price:", round(info["price"], 2), "| Stock:", info["stock"])

    item_name = input("Enter item name: ").strip().title()
    if item_name not in rest["menu"]:
        raise MenuItemError("Item not found in menu.")

    qty = int(input("Enter quantity: "))
    if qty <= 0:
        raise StockError("Quantity must be greater than 0.")

    item = rest["menu"][item_name]
    if qty > item["stock"]:
        raise StockError("Insufficient stock for this item.")

    total = round(item["price"] * qty, 2)
    if total < rest["min_order"]:
        raise MinimumOrderError("Order amount is below minimum order.")

    method = input("Payment method (wallet/cash): ").strip().lower()
    if method == "wallet":
        if total > cust["balance"]:
            raise PaymentError("Insufficient wallet balance.")
        cust["balance"] = round(cust["balance"] - total, 2)
    elif method != "cash":
        raise PaymentError("Invalid payment method.")

    item["stock"] = item["stock"] - qty
    orders.append({"customer": cname, "restaurant": rname, "item": item_name, "qty": qty, "total": total})
    print("Customer:", cname, "Restaurant:", rname, "Item:", item_name, "Quantity:", qty, "-> Order placed successfully")

except CustomerError as e:
    print("Customer error:", e)
except RestaurantError as e:
    print("Restaurant error:", e)
except MenuItemError as e:
    print("Menu error:", e)
except StockError as e:
    print("Stock error:", e)
except MinimumOrderError as e:
    print("Minimum order error:", e)
except DeliveryAreaError as e:
    print("Delivery error:", e)
except PaymentError as e:
    print("Payment error:", e)
except ValueError:
    print("Invalid number input.")
else:
    print("Total bill:", total)
finally:
    print("Order process ended.")
```
